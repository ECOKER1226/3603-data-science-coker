## 1. Headings and Text Styles

📌 **How the examples work:** each example shows **what you type** in a code block, followed by **what you get** when the cell renders. You can also double-click any rendered Markdown cell to see its source.

**You type:**
```markdown

### Heading 3

**Bold text**, *italic text*, ***bold and italic***, ~~strikethrough~~, `inline code`

---
```

**You get:**

# Heading 1

---

💡 **Headings are your table of contents.** In VS Code, open the **Outline** view (Explorer sidebar) to see every heading in the notebook and click one to jump straight to it. Well-organized headings make long notebooks easy to navigate.

### Paragraphs and Line Breaks
Markdown ignores single line breaks. Lines next to each other are joined into one paragraph. This is the most common "why is my text all on one line?" surprise.

**You type:**
```markdown
This line
and this line end up on the same line.

A blank line starts a new paragraph.

Use a <br> tag to force<br>a line break.
```

**You get:**

This line
and this line end up on the same line.

A blank line starts a new paragraph.

Use a <br> tag to force<br>a line break.

(Ending a line with **two spaces** also forces a line break, but you can't see the spaces, so `<br>` is easier to spot.)

## 4. LaTeX Math <a id="latex-math"></a>

Render math with LaTeX syntax: `$ ... $` for **inline** math (inside a sentence) and `$$ ... $$` for **block** math (centered on its own line).

**You type:**
```markdown
Einstein's famous equation is $E = mc^2$.

The sample mean is:
$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$
```

**You get:**

Einstein's famous equation is $E = mc^2$.

The sample mean is:
$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$

### LaTeX Cheat Sheet

| You type | You get | Meaning |
|---|---|---|
| `x^2`, `x^{10}` | $x^2$, $x^{10}$ | Superscript (use `{}` for more than one character) |
| `x_i`, `x_{ij}` | $x_i$, $x_{ij}$ | Subscript |
| `\frac{a}{b}` | $\frac{a}{b}$ | Fraction |
| `\sqrt{x}` | $\sqrt{x}$ | Square root |
| `\pm` | $\pm$ | Plus or minus |
| `\sum_{i=1}^{n}` | $\sum_{i=1}^{n}$ | Summation |
| `\bar{x}`, `\hat{y}` | $\bar{x}$, $\hat{y}$ | Mean, predicted value |
| `\alpha \beta \mu \sigma` | $\alpha \beta \mu \sigma$ | Greek letters |
| `\le \ge \ne \approx` | $\le \ge \ne \approx$ | Comparisons |

### ⚠️ The Dollar Sign Problem (and Escaping)
Because `$` starts math, ordinary prices can turn into a LaTeX mess:

The item costs $5 and the tax is $2.

Put a **backslash** `\` in front of a special character to show it literally:

**You type:**
```markdown

Show a literal \*asterisk\* and a \# hash sign.
```

**You get:**

The item costs \$5 and the tax is \$2.

## 5. Mixing Code and Markdown

`````python
# Example: create data in Python
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]
squares
`````

Now we explain the results in Markdown:
- The list `numbers` contained 1–5.
- The list `squares` contains each number squared.
- Mixing narrative and results is why notebooks are so powerful for data science reports.

### Generating Markdown from Code
The explanation above is **hard-coded**. If `numbers` changes, the text is now wrong. You can have Python **write** the Markdown for you by using an f-string to put the real values in. (This is the same idea as the `%%markdown` cell magic from notebook 01, but it can use your variables.)

`````python
from IPython.display import Markdown, display

display(Markdown(f"""
The list `squares` has **{len(squares)}** values.
Their sum is **{sum(squares)}** and the largest is **{max(squares)}**.
"""))
`````

## The Pythagorean Theorem

The Pythagorean Theorem states that in any right triange, the square of the length of the hypotenuse ($c) equals the sum of the squares of the length of the two legs ($a and $b).

- Legs ($a and $b): The two shorter sides that meet to form a 90-degree right angle.
- Hypotenuse ($c): The longest side of the right triangele, located directly opposite the right angle.

$$a^2 + b^2 = c^2$$
